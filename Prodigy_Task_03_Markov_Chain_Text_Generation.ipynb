```python
# Markov Chain Text Generation

import random
import re

print("Markov Chain Text Generator")
```

```python
text = """
Artificial Intelligence is transforming the world.
Machine Learning helps computers learn from data.
Deep Learning is a part of Machine Learning.
Artificial Intelligence is used in healthcare and education.
Machine Learning is useful for prediction and analysis.
Technology makes our daily life easier.
Deep Learning powers many modern AI applications.
Artificial Intelligence is growing rapidly.
"""

print(text)
```

```python
# Create Markov Chain model

words = re.findall(r'\b\w+\b', text.lower())

markov_chain = {}

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]

    if current_word not in markov_chain:
        markov_chain[current_word] = []

    markov_chain[current_word].append(next_word)

print("Markov Chain created successfully!")
print(markov_chain)
```

```python
def generate_text(start_word, length=20):
    current_word = start_word.lower()
    result = [current_word]

    for _ in range(length - 1):
        if current_word not in markov_chain:
            break

        next_word = random.choice(markov_chain[current_word])
        result.append(next_word)
        current_word = next_word

    return " ".join(result)


generated_text = generate_text("artificial", 20)

print("Generated Text:")
print(generated_text)
```

```python
print("Output 1:", generate_text("machine", 15))
print("Output 2:", generate_text("deep", 15))
print("Output 3:", generate_text("technology", 15))
```
